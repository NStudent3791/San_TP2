```python
import numpy as np
import msicpe.san as san
from plotly import express as px
import scipy.signal as signal
import pandas as pd

ecg_signal_dict = np.load('ecg_signal.npz')
ecg_s = ecg_signal_dict['s'] 
ecg_t = ecg_signal_dict['t']

ecg_nu, ecg_S = san.trans_fourier(ecg_s, ecg_t)
ecg_S = np.abs(ecg_S)

ds = pd.DataFrame({'ecg_s' : ecg_s, 'ecg_t':ecg_t})
dS = pd.DataFrame({'ecg_nu':ecg_nu, 'ecg_S':ecg_S})

fig_ecg = px.line(ds, x = 'ecg_t', y = 'ecg_s')
fig_spectre_ecg = px.line(dS, x = 'ecg_nu', y = 'ecg_S')

fig_ecg.show()
fig_spectre_ecg.show()
```

```python
signal_dict = np.load('tuning_fork_440Hz.npz')
s = signal_dict['s'] 
t = signal_dict['t']

nu, S = san.trans_fourier(s, t)
S = np.abs(S)

ds = pd.DataFrame({'s' : s, 't':t})
dS = pd.DataFrame({'nu':nu, 'S':S})

fig = px.line(ds, x = 't', y = 's')
fig_spectre = px.line(dS, x = 'nu', y = 'S')

fig.show()
fig_spectre.show()
```
